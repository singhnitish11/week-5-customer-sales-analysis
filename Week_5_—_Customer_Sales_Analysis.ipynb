```python
# Identify possible customer ID and sales columns

customer_id_candidates = [
    col for col in sales_df.columns
    if any(word in col.lower()
           for word in ["customer_id", "customerid", "customer", "client_id", "client", "user_id"])
]

sales_value_candidates = [
    col for col in sales_df.columns
    if any(word in col.lower()
           for word in ["sales", "sale", "revenue", "amount", "total", "price"])
]

print("Possible Customer ID Columns:")
print(customer_id_candidates)

print("\nPossible Sales Value Columns:")
print(sales_value_candidates)
```

## Customer-Level Sales Aggregation

The sales data is grouped by customer to calculate total sales,
average sales, and number of transactions for each customer.

```python
if customer_id_candidates and sales_value_candidates:

    customer_id_col = customer_id_candidates[0]
    sales_value_col = sales_value_candidates[0]

    sales_df[sales_value_col] = pd.to_numeric(
        sales_df[sales_value_col],
        errors="coerce"
    )

    customer_sales = (
        sales_df
        .groupby(customer_id_col)[sales_value_col]
        .agg(
            total_sales="sum",
            average_sales="mean",
            transaction_count="count"
        )
        .sort_values("total_sales", ascending=False)
    )

    print("Customer-level sales analysis:")
    display(customer_sales.head(10))

else:
    print("Customer ID or sales value column was not found.")
    print("Please check the column names before performing customer-level analysis.")
```

```python
if categorical_cols and numeric_cols:

    category_col = categorical_cols[0]
    value_col = numeric_cols[0]

    pivot_summary = pd.pivot_table(
        sales_df,
        index=category_col,
        values=value_col,
        aggfunc=["sum", "mean", "count"],
        fill_value=0
    )

    print("Pivot Table Summary:")
    display(pivot_summary)

else:
    print("Pivot table requires at least one categorical")
    print("column and one numeric column.")
```

## Multiple-Condition Filtering

Pandas supports multiple conditions using:

- AND (`&`): Both conditions must be true.
- OR (`|`): At least one condition must be true.

We will filter the sales data using available numeric columns.

```python
if len(numeric_cols) >= 2:

    col1 = numeric_cols[0]
    col2 = numeric_cols[1]

    median1 = sales_df[col1].median()
    median2 = sales_df[col2].median()

    filtered_and = sales_df[
        (sales_df[col1] > median1) &
        (sales_df[col2] > median2)
    ]

    print("Records satisfying both conditions:")
    display(filtered_and.head(10))

else:
    print("At least two numeric columns are needed")
    print("for this AND-condition example.")
```

## OR Condition Filtering

The OR operator (`|`) selects records that satisfy
at least one of the specified conditions.

```python
if len(numeric_cols) >= 2:

    filtered_or = sales_df[
        (sales_df[col1] > median1) |
        (sales_df[col2] > median2)
    ]

    print("Records satisfying at least one condition:")
    display(filtered_or.head(10))

else:
    print("OR-condition analysis requires two numeric columns.")
```

```python
numeric_data = sales_df.select_dtypes(
    include="number"
)

if numeric_data.shape[1] >= 2:

    correlation = numeric_data.corr()

    plt.figure(figsize=(8, 6))

    image = plt.imshow(
        correlation,
        aspect="auto",
        vmin=-1,
        vmax=1
    )

    plt.colorbar(image, label="Correlation")

    plt.xticks(
        range(len(correlation.columns)),
        correlation.columns,
        rotation=45,
        ha="right"
    )

    plt.yticks(
        range(len(correlation.columns)),
        correlation.columns
    )

    plt.title("Numeric Data Correlation Heatmap")
    plt.tight_layout()

    plt.savefig(
        "visualizations/correlation_heatmap.png",
        dpi=150,
        bbox_inches="tight"
    )

    plt.show()

else:
    print("At least two numeric columns are required.")
```
